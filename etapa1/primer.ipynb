```python
df_csv = pd.read_csv('raw_perfiles_usuarios.csv')
display(df_csv.head())

```

```python
df_csv = df_csv.dropna()

df_csv['row_id'] = df_csv['row_id'].str.upper()
df_csv['user_id'] = df_csv['user_id'].str.upper()
df_csv['pais'] = df_csv['pais'].str.upper()
df_csv['dispositivo'] = df_csv['dispositivo'].str.upper()
df_csv['nivel_verificacion'] = df_csv['nivel_verificacion'].str.upper()
df_csv['version_api'] = df_csv['version_api'].str.upper()


rows = df_csv[df_csv["edad_usuario"] > 0].index
df_csv.drop(rows, inplace=True)

rows = df_csv[df_csv["seguidores"] > 0].index
df_csv.drop(rows, inplace=True)

rows = df_csv[df_csv["posts_totales"] > 0].index
df_csv.drop(rows, inplace=True)

rows = df_csv[df_csv["tasa_interaccion"] > 0].index
df_csv.drop(rows, inplace=True)

rows = df_csv[df_csv["hora_pico_actividad"] > 0].index
df_csv.drop(rows, inplace=True)

df_csv = df_csv[(df_csv['dispositivo'] != 'N/A') & (df_csv['nivel_verificacion'] != 'N/A') & (df_csv['seguidores'] != 'N/A') & (df_csv['siguiendo'] != 'N/A') & (df_csv['posts_totales'] != 'N/A') & (df_csv['tasa_interaccion'] != 'N/A') & (df_csv['hora_pico_actividad'] != 'N/A') & (df_csv['version_api'] != 'N/A')]

df_csv.drop_duplicates(subset=['row_id', 'user_id'], keep='last')

df_csv['fecha_creacion'] = pd.to_datetime(df_csv['fecha_creacion'], format='%d/%m/%Y')


display(df_csv.head())
print(len(df_csv))

```
