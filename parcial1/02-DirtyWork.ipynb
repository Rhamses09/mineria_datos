```python
import random

random_list = [random.random() for _ in range(5_000_000)]

print("Random list length: ", len(random_list))
```

```python

total = 0

for i in range(len(random_list)):
    total += random_list[i]
    
total = total / len(random_list)

suma = 0

for i in range(len(random_list)):
    suma += (random_list[i] - total) * (random_list[i] - total)

final = suma / len(random_list)

sd = pow(final, 0.5)

print(sd)
```

```python
import numpy as np 

arr = np.array(random_list)

print(f'Std: {arr.std()}')
```
