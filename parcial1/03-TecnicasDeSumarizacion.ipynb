```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
skew = df['Age'].skew()
print(f'{skew}')
kurt = df['Fare'].kurt()
print(f'{kurt}')
cross = pd.crosstab(df['Pclass'], df['Survived'], margins=True)
print(f'{cross}')
```

Asimetría (Skewness): Investiguen cómo aplicar la función .skew() en la columna Age. Si el resultado es positivo o negativo, ¿qué significa físicamente sobre la edad de los pasajeros en el barco? (¿Había más jóvenes o más ancianos?).

Viendo que da un resultado de 0.3891, esto significa que hay mas personas jovenes, ya que la cola de distribuicion se estira a la derecha, osea edades mayores, lo que nos dice que la mayoria esta en la izquierda que son edades jovenes

Curtosis (Kurtosis): Investiguen cómo aplicar la función .kurt() en la columna Fare. ¿Qué significa este número sobre el "grosor" de las colas de la distribución? ¿Confirma esto la existencia de los outliers que encontramos hoy en clase?

Se obtiene el valor de 33.398, que es un numero alto positivo, esto significa que la distribucion tiene colas pesadas o gruesas. Confirmando una existencia de outliers, inficando que hay datos extremos o alejados del promedio

Tablas Cruzadas (Cross-Tabulation): Investiguen la función pd.crosstab(). Generen una tabla cruzada entre la clase del pasajero (Pclass) y si sobrevivió (Survived). Esta técnica de sumarización matricial es el equivalente a una Tabla Dinámica en Excel. ¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?

Al momento de cruzar Pclass con Survived, el patron dice que la clase social pudo haber tenido algo que ver con salvarse, ya que los pasajeros de tercera clase fueron los que tuvieron mayores bajas hablando de porcentaje, a comparacion de los de primera clase, donde sobrevivieron la mayoria, en segunda clase fue casi la mitad, mostrando que pasajeros de una clase alta tenia estadisticamente una mayor probabilidad de sobrevivir
