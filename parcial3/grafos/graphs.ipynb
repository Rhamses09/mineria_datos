```python
import pandas as pd
import numpy as np

df = pd.read_csv('vuelos_10k.csv')


df = df.dropna()
df['Origen'] = df['Origen'].str.upper() 
df['Destino'] = df['Destino'].str.upper()
df = df[df['Origen'] != df['Destino']]
df = df[df['Distancia_km'] > 0]
df.head()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

G = nx.from_pandas_edgelist(df, source="Origen", target="Destino", edge_attr="Distancia_km")

print(len(G.nodes()))
print(len(G.edges()))
```

```python
matriz = nx.adjacency_matrix(G).todense()
print(matriz)
print(matriz.nbytes)
print(df.memory_usage(deep=True))
```

¿Cuántas veces más pesada es la Matriz en comparación con la Lista de Aristas y por qué sucede esto?

R: La matriz siendo 84 megas, procesa todas las conexiones que tiene junto con todos los otros nodos, siendo casi n^2 si se tuvieran conexiones de cada nodo con todos, como la tabla de adyecencia siendo redundante, y la tabla simplemente guarda valores pero sin establecer conexiones entre si en el dataframe.
