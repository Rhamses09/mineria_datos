```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

duplicados = df.duplicated(subset=['App']).sum()
print("Duplicados encontrados:", duplicados)

df.drop_duplicates(subset=['App'], keep='first', inplace=True)
print("Total de filas tras purgar duplicados:", len(df))
```

```python
df = df[df['Installs'] != 'Free']

df['Installs'] = df['Installs'].str.replace('+', '', regex=False).str.replace(',', '', regex=False).astype(int)

print("Promedio real de descargas:", df['Installs'].mean())
```

```python
df['Price'] = df['Price'].str.replace('$', '', regex=False).astype(float)

print(df['Price'].describe())
```

Pregunta C: La columna Price tiene valores como "$4.99" y "0". Apliquen la misma lógica arquitectónica que en la Pregunta B: eliminen el símbolo de dólar $ y conviertan la columna usando .astype(float).

Respuesta: Ahora usa valores flotantes reales

```python
print("Precio máximo:", df['Price'].max())

print("\nApps con precio mayor a 200 dolares:")
print(df[df['Price'] > 200][['App', 'Price']])

df = df[df['Price'] < 50]
print("\nTotal de registros con Price < 50:", len(df))

df.to_csv('playstore_limpio.csv', index=False)
print("Dataset limpio guardado exitosamente en 'playstore_limpio.csv'.")
```
